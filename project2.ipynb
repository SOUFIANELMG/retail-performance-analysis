```python
def build_kpi(group_col):

    kpi = valid_orders.groupby(group_col).agg(
        total_orders=("order_id", "count"),
        total_revenue=("realized_revenue", "sum"),
        total_profit=("profit_amount", "sum")
    )

    delivered = valid_orders[
        valid_orders["status_clean"] == "DELIVERED"
    ].groupby(group_col)["order_id"].count()

    kpi["delivered_orders"] = delivered

    kpi["AOV"] = (
        kpi["total_revenue"] /
        kpi["delivered_orders"]
    )

    known = known_profit_orders.groupby(group_col).agg(
        known_revenue=("realized_revenue", "sum"),
        known_profit=("profit_amount", "sum")
    )

    kpi["profit_margin"] = (
        known["known_profit"] /
        known["known_revenue"]
    ) * 100

    return kpi
```

```python
region_kpi = build_kpi("region_clean")

channel_kpi = build_kpi("sales_channel_clean")

payment_kpi = build_kpi("payment_method_clean")
```
